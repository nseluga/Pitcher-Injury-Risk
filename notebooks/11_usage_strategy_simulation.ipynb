## Planned Analyses

### Pitch Mix Simulations

### Role Simulations

2. Role transition simulation: estimated risk change for starters converting to relief
3. Staff-level workload redistribution: minimize total staff IL days

## Simulation Methodology
All simulations are counterfactual: we modify specific features in a pitcher's
actual feature vector (holding all other features constant) and re-run model
inference. This assumes the model has learned reliable conditional relationships —
a strong assumption that must be acknowledged.

**Important limitation:** These simulations assume ceteris paribus (all else equal).
In reality, changing pitch count or pitch mix may affect performance, which in turn
affects sequencing decisions, workload distribution, etc. Causal claims require
stronger evidence than model-based simulation alone.
